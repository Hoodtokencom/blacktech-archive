```python
# STEP 1/7 - inputs: unpack the embedded avatar + voice, transcode to a clean 16 kHz WAV.
import base64, os, subprocess
os.makedirs('/content/inputs', exist_ok=True)
AVATAR = '/content/inputs/avatar.jpg'
AUDIO_MP3 = '/content/inputs/voice.mp3'
AUDIO = '/content/inputs/voice.wav'
for path, blob, what in ((AVATAR, AVATAR_B64, 'avatar'), (AUDIO_MP3, VOICE_B64, 'voice')):
    open(path, 'wb').write(base64.b64decode(''.join(blob.split())))
    print(f'{what:6s} -> {path}  {os.path.getsize(path)/1024:.1f} KB')
subprocess.run(['ffmpeg', '-y', '-loglevel', 'error', '-i', AUDIO_MP3, '-ar', '16000', '-ac', '1', AUDIO],
               check=True)
print(f'voice  -> {AUDIO}  {os.path.getsize(AUDIO)/1024:.1f} KB  (16 kHz mono wav)')
assert os.path.getsize(AVATAR) > 5000 and os.path.getsize(AUDIO) > 5000
print('\n✅ STEP 1/7 OK - inputs ready')
```

```python
# STEP 2/7 - is the GPU real?
import subprocess, torch
print(subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv,noheader'],
                     capture_output=True, text=True).stdout.strip())
assert torch.cuda.is_available(), (
    '❌ STEP 2/7 FAILED - NO GPU. Runtime → Change runtime type → T4 GPU, then Run all again')
p = torch.cuda.get_device_properties(0)
print(f'torch {torch.__version__} | numpy {__import__("numpy").__version__} | GPU: {p.name} | {p.total_memory/1e9:.1f} GB VRAM')
print('\n✅ STEP 2/7 OK - GPU live')
```

```python
# STEP 3/7 - install SadTalker (idempotent: skips if already done in this runtime).
# NOTE: torch / torchvision / torchaudio / numpy lines are FILTERED OUT of requirements.txt so
# Colab's CUDA torch and the system numpy survive. (SadTalker's requirements have no torch pin.)
import os, subprocess, sys
if not os.path.isdir('/content/SadTalker/.git'):
    subprocess.run(['rm', '-rf', '/content/SadTalker'], capture_output=True)
    subprocess.run(['git', 'clone', '-q', 'https://github.com/OpenTalker/SadTalker.git',
                    '/content/SadTalker'], check=True)
    print('cloned SadTalker')
else:
    print('SadTalker already present - skipping clone')
os.chdir('/content/SadTalker')
subprocess.run(['apt-get', '-qq', 'install', '-y', 'libgl1', 'libglib2.0-0'], capture_output=True)
if not os.path.exists('/content/.deps_ok'):
    lines = [l for l in open('requirements.txt').read().splitlines()
             if l.strip() and not l.lower().lstrip().startswith(
                 ('torch', 'torchvision', 'torchaudio', 'numpy'))]
    open('req_nt.txt', 'w').write('\n'.join(lines))
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'req_nt.txt'], check=True)
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'gfpgan', 'facexlib'], check=True)
    open('/content/.deps_ok', 'w').write('1')
    print('deps installed')
else:
    print('deps already installed - skipping')
import torch
print('torch now:', torch.__version__, '| cuda:', torch.cuda.is_available())
assert torch.cuda.is_available(), '❌ STEP 3/7 FAILED - pip clobbered CUDA torch. Restart runtime, Run all.'
print('\n✅ STEP 3/7 OK - deps + CUDA torch intact')
```

```python
# STEP 4/7 - THE FIX. SadTalker (2023) imports names that numpy 2.0 DELETED, so the script
# dies on import before any face work:  np.VisibleDeprecationWarning (and np.float elsewhere).
# Repair them in the checked-out source. Idempotent + safe to re-run.
import glob, pathlib, re
changed = []
for p in glob.glob('/content/SadTalker/**/*.py', recursive=True):
    f = pathlib.Path(p)
    t = f.read_text()
    n = t.replace('category=np.VisibleDeprecationWarning',
                  'category=getattr(np, "exceptions", np).VisibleDeprecationWarning')
    n = re.sub(r'\bnp\.float\b(?!\d)', 'np.float64', n)
    if n != t:
        f.write_text(n)
        changed.append(p)
print('numpy-2 patches applied to:', len(changed), 'file(s)')
for c in changed:
    print('  ', c.replace('/content/SadTalker/', ''))
# prove the exact line that crashed v3 is now correct
crash = pathlib.Path('/content/SadTalker/src/face3d/util/preprocess.py')
line = [l for l in crash.read_text().splitlines() if 'VisibleDeprecationWarning' in l]
print('line now:', line[0].strip() if line else '(not found!)')
assert line and 'VisibleDeprecationWarning' in line[0] and 'np.VisibleDeprecationWarning' not in line[0]
print('\n✅ STEP 4/7 OK - numpy 2 compatibility fixed')
```

```python
# STEP 5/7 - patch basicsr: it imports the removed torchvision.transforms.functional_tensor
# on Colab's torchvision, which makes inference.py die on import. Safe if already correct.
import glob, pathlib
n = 0
for p in glob.glob('/usr/local/lib/python3*/dist-packages/basicsr/data/degradations.py'):
    f = pathlib.Path(p)
    t = f.read_text()
    if 'functional_tensor' in t:
        f.write_text(t.replace(
            'from torchvision.transforms.functional_tensor import rgb_to_grayscale',
            'from torchvision.transforms.functional import rgb_to_grayscale'))
        n += 1
print('basicsr files patched:', n)
print('\n✅ STEP 5/7 OK')
```

```python
# STEP 6/7 - checkpoints (~2.4 GB first time; instant if already present).
import os, subprocess
os.chdir('/content/SadTalker')
subprocess.run(['bash', 'scripts/download_models.sh'])
need = ['checkpoints/SadTalker_V0.0.2_256.safetensors', 'gfpgan/weights/GFPGANv1.4.pth']
missing = [f for f in need if not os.path.exists(f)]
assert not missing, f'❌ STEP 6/7 FAILED - missing {missing}'
print('\n✅ STEP 6/7 OK - checkpoints present')
```

```python
# STEP 7/7 - RENDER. Two attempts: best quality first, then a plain fallback, so one
# failure cannot end the run. Full log goes to a file; a short tail is printed (fits one photo).
# Nothing is uploaded anywhere.
import glob, os, subprocess, sys
os.chdir('/content/SadTalker')

BASE = [sys.executable, 'inference.py', '--driven_audio', AUDIO, '--source_image', AVATAR]
attempts = [
    ('crop + gfpgan enhancer', BASE + ['--result_dir', '/content/out1',
                                       '--still', '--preprocess', 'crop', '--enhancer', 'gfpgan', '--size', '256']),
    ('plain crop (no enhancer)', BASE + ['--result_dir', '/content/out2',
                                         '--still', '--preprocess', 'crop', '--size', '256']),
]
logs, vids = [], []
for i, (label, cmd) in enumerate(attempts, 1):
    print(f'\\n===== ATTEMPT {i}: {label} =====')
    print('RUN:', ' '.join(cmd))
    p = subprocess.run(cmd, capture_output=True, text=True)
    log = p.stdout + '\\n\\n===== STDERR =====\\n' + p.stderr
    logs.append(f'===== ATTEMPT {i} ({label}) =====\\n' + log)
    vids = sorted(glob.glob(cmd[cmd.index('--result_dir') + 1] + '/**/*.mp4', recursive=True))
    print(f'attempt {i}: exit={p.returncode}  mp4s={len(vids)}')
    if vids:
        print(f'\u2705 SUCCESS on attempt {i}')
        break

log = '\\n\\n'.join(logs)
open('/content/render_log.txt', 'w').write(log)
print('\\nfull log saved to /content/render_log.txt (' + str(len(log)) + ' chars)')

if not vids:
    # A Python traceback keeps its cause in the LAST lines, so the tail is the whole story.
    print('\\n===== LAST 22 LINES OF OUTPUT (the error is in here) =====')
    for line in log.strip().splitlines()[-22:]:
        print(line)
    print('\\n\u274c STEP 7/7 FAILED - no mp4 produced. Photo the 22-line tail above.')
else:
    print('\\n✅ STEP 7/7 OK - clip rendered')
```

```python
# Report + auto-download (interactive only).
import glob, os, shutil
vids = sorted(glob.glob('/content/out*/**/*.mp4', recursive=True), key=os.path.getmtime)
print('videos produced:', len(vids))
for v in vids:
    print(' ', v, f'{os.path.getsize(v)/1024:.0f} KB')
if not vids:
    print('⚠ no mp4 - photo the error tail printed at the end of STEP 7/7.')
else:
    try:
        from google.colab import files
        shutil.copy(vids[-1], '/content/talking_head.mp4')
        files.download('/content/talking_head.mp4')
        print('✅ downloading talking_head.mp4')
    except Exception as e:
        print('download the file directly from the Colab file browser:', e)
```
